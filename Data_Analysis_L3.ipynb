# Arrays

We define an array by specifying that it is an NumPy array
```python
a = np.array([1,2,3])
```
The indexing of arrays start from 0.

That is, a[0] = 1, a[1] = 2 and a[2] = 3. a[3] is undefined.

Similarly, we can define a matrix by adding '[]'. For example, to define the following matrix

$$
\begin{equation}
b =
\begin{pmatrix}
  1 & 0 & -5 \\
  -2.5 & 7.3 & 4.6 
\end{pmatrix}
\end{equation}
$$

in Python, we write

````python
b = np.array([[1,0,-5],[-2.5,7.3,4.6]])
````

````python
b
````

Indexing starts from the outer brackets, that is

b[0,0] = 1, b[0,1] = 0, b[1,1] = 7.3

Higher dimensional array can be defined by adding more '[]'.
For example, we can define a zero matrix with dimension 2 x 4 x 3

````python
d = np.zeros((2,4,3))
````

````python
d
````

````python
b.shape[0]
#In array logic, axis 0 means we are focusing on rows.
````

````python
b.shape[1]
#In array logic, axis 1 means we are focusing on columns.
````

## Indexing and Slicing

As mentioned above, the indexing starts from 0.
The ':' symbol is important in indexing, it gives all values in the array.
For example, with

````python
a = np.array([1,2,3])
````

````python
a[:]
````

Use Matrix b as another example: 

b[:,0] slice out all entries in the first column, giving [1,-2.5] 

b[1,:] slice out all entries in the second row, giving [-2.5,7.3,4.6]

````python
b[:,0]
````

````python
b[1,:]
````

We can also slice part of matrix or array by using 'Start:Stop' notation. For example:

````python
b
````

````python
b[1:,:2]
````

````python
b[:2,1:]
````

````python
b[:,1:]
````

## Stack

We can actually stack/combine arrays/matrices together.

````python
A = np.array([[1,1],[0,0]])
A
````

````python
B = np.array([[2,2,],[3,3]])
B
````

````python
np.vstack((A,B))
````

````python
np.hstack((A,B))
````

## Arithmetic

Basic operations are element-wise
- \+ addition
- \- subtraction
- \* multiplication
- / division
- \** power

For example, to take the square of every element in array u:

````python
u = np.array([1,2,3,4,5])
````

````python
v = u ** 2
v
````

The square root can be computed in two ways

````python
v ** 0.5
````

````python
np.sqrt(v)
````

Note that, you can sum over all entries in an array by

````python
np.sum(v)
````

To get the real part of an array, we have

````python
v.real
````

## Special Arrays

### Equally Spaced Array

By specifying starting value, ending value (minus increment) and the increment, andequally spaced array is defined by 'arange'

You can also specify the number of entries only, then an array of integer starting from 0 is defined

````python
p = np.arange(-5,5,0.1)
q = np.arange(5)
````

````python
p
````

````python
q
````

We can also apply 'linspace' where two endpoints and length of array is specified.

````python
np.linspace(0,25,100)
````

# Matrix Operations

## Matrix Multiplication

To multiply two matrices x and y, we either use the 'dot' function or simply '@'

````python
x = np.array([[3,3],[4,4]])
y = np.array([[2,5],[0,1]])
````

````python
np.dot(x,y)
````

````python
x @ y
````

## Inverse

To compute the inverse of a squared matrix, we need to import the inverse function from linear algebra library 'linalg' in NumPy

````python
from numpy.linalg import inv
````

````python
inv(y)
````

# Statistics

Suppose we have a 2 by 3 matrix

````python
C = np.array([[1,0,2],[3,5,6]])
````

Mean of all entries is given by

````python
np.mean(C)
````

````python
#Alternative
C.mean()
````

We can also compute mean along rows or columns by specifying the axis

````python
C.mean(axis = 0)
````

This gets the means of the three columns.

````python
C.mean(axis = 1)
````

And this gives the means of the two rows.

For standard deviation, the function is

````python
C.std()
````

And for maximun

````python
C.max()
````
