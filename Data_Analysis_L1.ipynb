## List

Lists are used to store multiple items in a single variable.

```python
x = [2, 4, 6, 8, 10]
y = [6,5,5,5,6,6,5,6]
```

Lists are indexed, the first entry is indexed with 0, the second is indexed 1 etc.

```python
x[0]
```

You can also add items or remove items to the list.

```python
x = x + [6]
print(x)
```

list.remove will remove the item in the list, when there are duplicated items in the list, the function will remove the one with smaller index. To remove specified index item, use list.pop.

```python
x.remove(6)
print(x)
```

```python
y.pop(6)
print(y)
```

## Range

The range() function returns a sequence of numbers, starting from 0 by default, and increments by 1 (by default), and stops before a specified number.

```python
z = range(6)
for n in z:
    print(n)
```

The syntax of range() goes like this: range(start, stop, step). Therefore, we can:

```python
w = range(4, 20, 2)
for m in w:
    print(m)
```

Note that this function will generate sequence up to n-1 for range(n)

## Boolean

Booleans represent one of two values: True or False.

Almost any value is evaluated to True if it has some sort of content.

Any string is True, except empty strings.

Any number is True, except 0.

Any list, tuple, set, and dictionary are True, except empty ones.

```python
print(bool(0))
print(bool(False))
print(bool(15))
print(bool("Hello, World!"))
```

Logic relationships are also Boolean type.

```python
print(10 > 9)
print(10 == 9)
print(10 < 9)
```

## Dictionary

Dictionaries are used to store data values in key:value pairs.

A dictionary is a collection which is ordered, changeable and do not allow duplicates.

```python
dict_example = {
  "brand": "McDonald",
  "food": "hamburger",
  "price": 28
}
print(dict_example)
```

Dictionaries will be really useful in later lessons while we handle datasets.
