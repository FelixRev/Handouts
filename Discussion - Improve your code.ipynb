# Spot the error

Before you can improve your code, you must recognize what’s wrong. Errors in code are not just obstacles—they are learning opportunities.

**Common sources of errors include:**
- **Typographical mistakes** (e.g., variable or function name misspelled)
- **Incorrect assumptions** about library functions or language features
- **Misunderstanding data types and structures**
- **Logic errors** that don’t trigger exceptions but yield wrong results

Let’s examine typical mistakes using `numpy`.

````python
import numpy as np

# Example 1: Typo in function name
try:
    result = np.sqaure_root(9)
except Exception as e:
    print("Error (typo in function name):", e)

# Example 2: Using a function that doesn't exist in numpy
try:
    result = np.powerof(2, 3)
except Exception as e:
    print("Error (non-existent function):", e)

# Example 3: Logic error - correct function, but wrong usage
try:
    result = np.sqrt([4, -1, 9])  # sqrt of negative number in real domain
    print("Result:", result)
except Exception as e:
    print("Error (logic/domain error):", e)
````

**Why Do These Errors Occur?**

- **Typo in Function Name:**  
  `sqaure_root` should be `sqrt`. Python is case- and spelling-sensitive.
- **Non-existent Function:**  
  `powerof` does not exist in numpy; the intended function is `power`.
- **Logic/Domain Error:**  
  `np.sqrt([4, -1, 9])` is valid syntax, but mathematically, square roots of negative numbers are not defined in the set of real numbers, so numpy returns `nan` for invalid inputs.


**Why do these mistakes persist?**  
  Even experienced developers are prone to such oversights. The key is to quickly identify and correct them, developing a habit of reading tracebacks and understanding error messages.
**How to improve?**  
  - Regularly consult documentation even for familiar libraries.
  - Write small test cases to verify assumptions.
  - Use IDE or notebook autocomplete features to reduce typos.

# Cons of Debugging Your Code Through AI Assistance

AI tools (like ChatGPT, Copilot, etc.) are increasingly used for code help. While they can save time, there are pitfalls:

### 1. Outdated Knowledge Base
- AI models are trained on past data, so their suggestions may reference deprecated functions or outdated practices.
- **Example:** AI might suggest `numpy.asscalar()` which is removed in recent numpy versions.

### 2. Repetitive or Generic Answers
- AI might provide template responses ("Check your syntax", "Update your library") that do not address the specifics of your code, especially for nuanced bugs.

### 3. Over-Complex Solutions
- Sometimes, AI recommends unnecessary libraries or convoluted workarounds.
- This bloats your codebase and can introduce dependency conflicts.

### 4. False Sense of Understanding
- Relying on AI can lead to "copy-paste" coding, where you use code without understanding. This hinders your long-term growth as a developer.

### 5. Limited Context Awareness
- AI might not be aware of your project's full codebase, environment, or constraints, so their advice could inadvertently break other parts of your code.

# Manual Debugging Methods

Developing strong manual debugging skills is crucial. Here are several strategies:

### 1. Consult Documentation and Community Resources
- Always check the official documentation for the function or library you are using.
- Use Stack Overflow, GitHub Issues, and other developer forums for real-world examples and known issues.

### 2. Search for Updated Information
- Google the function name and error message.
- Specify versions (e.g., "numpy sqrt negative number error version 1.26").

### 3. Check for Naming Conflicts
- Ensure that your variable, function, and file names do not shadow library functions.
- For example, naming a script `numpy.py` can break imports from the actual numpy library.

### 4. Use Print Statements and Logging
- Insert `print()` statements to inspect variable values and flow.
- For larger projects, use Python’s built-in `logging` module for more control.

### 5. Use Interactive Debuggers
- Use Python’s `pdb` or `%debug` in Jupyter to step through code, check variable states, and set breakpoints.

### 6. Write Minimal, Reproducible Examples
- Isolate the problematic code in a small script or notebook cell.
- This helps you focus on the bug and makes it easier to get help from others.

### 7. Test Assumptions with Unit Tests
- Write small tests to confirm that functions behave as expected.
- Use libraries like `pytest` or Python’s built-in `unittest`.
