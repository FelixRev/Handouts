# Line plot

We input an array of x-coordinates and an array of the corresponding y-coordinates.

The 'plot' function will plot the ordered pairs and join them by a line.

Example

To plot a parabola, we simply define the x-vector of dicretized coordinates and the y-vector as the square of x.

Then, the command 'plt.plot(x,y)' gives the graph of y against x.

````python
x = np.arange(10)
y = x ** 2
plt.plot(x,y)
````

## Line style

For the graph plotted, we can change the 
- line style
- color
- marker
- label

For line style,

'-' is solid line

'--' is dashed line

':' is dotted line

````python
plt.plot(x,y,':')
````

````python
plt.plot(x,y,'--')
````

## Color

While color is described in alphabets:

- 'b' is blue
- 'r' is red
- 'g' is green
- 'k' is black

## Marker

We can also mark the location of the point by specifying the style of the marker,

- 'x' for a cross
- 'o' for circle, etc.

````python
plt.plot(x,y,'ro--')
````

## Label

We can label the graph with different commands:

````python
plt.plot(x,y)
plt.xlabel("x")
plt.ylabel("y")
plt.title('My First Plot')
````

## Others

We can also specify the range of the axes by
``` python
plt.xlim([0,10])
plt.ylim([0,100])

We can also add legen to the graph with 'legend'. The location of label is specified by 'loc'.

````python
f1 = x ** 2
f2 = x ** 2 + 2 * x + 10
plt.plot(x, f1, 'b-', label = 'f1')
plt.plot(x, f2, 'r:', label = 'f2')
plt.xlabel("x")
plt.ylabel("y")
plt.title('My First Plot')
plt.xlim([0,10])
plt.ylim([0,100])
plt.legend()
````

# Scatter plot

The pyplot function 'scatter' provides a scatter plot.

Similar to the syntax in plotting a line, we specify the x- and y- coordinates.

Later when we work of classification problems, we mya like to distinguish classes of data points.

Classes of points can be distinguished bby 2 separate lines of commands.

Example

````python
p1 = np.array([[1,1],[1,2],[2,2]])
p2 = np.array([[3,4],[5,3]])
plt.scatter(p1[:,0], p1[:,1], color = 'r', marker = 'o')
plt.scatter(p2[:,0], p2[:,1], color = 'b', marker = 'x')
````

# Histogram

To create a histogram, we apply the 'hist' function.

And the number of bars is specified by 'bine'.

Example

Recall that we can generate an array of random numbers from a specific distribution. For example, with the 'rand' command, we obtain samples from an uniform distribution. In the following example, we plot the histogram of 1000 random numbers drawn from an uniform distribution.

# Surface plot

To plot a 3-dimenstional surface, we need a special toolkit.

````python
from mpl_toolkits.mplot3d import Axes3D
````

Notice that the functional values should form a matrix.

The 'meshgrid' function in NumPy is useful for this purpose.

Example

````python
x = np.arange(-10, 10)
y = np.arange(-10, 10)
xx, yy = np.meshgrid(x, y)
zz = xx**2 + yy**2
figure = plt.figure()
ax = figure.add_subplot(111, projection='3d')
ax.plot_surface(xx, yy, zz)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
````
