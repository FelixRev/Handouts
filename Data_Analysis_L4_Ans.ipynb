## Exercise 1

```python
import numpy as np
import matplotlib.pyplot as plt

# Create x values
x = np.arange(-10, 10)
# Calculate y values
y = 2 * x + 3

# Create the plot
plt.figure()
plt.plot(x, y, label='y = 2x + 3')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Plot of y = 2x + 3')
plt.show()
```

## Exercise 4

```python
from mpl_toolkits.mplot3d import Axes3D

# Create data
x = np.arange(-20, 20)
y = np.arange(-20, 20)
xx, yy = np.meshgrid(x, y)
zz = xx**2 + yy**2

# Create a 3D surface plot
figure = plt.figure()
ax = figure.add_subplot(111, projection='3d')
ax.plot_surface(xx, yy, zz, cmap='viridis')

# Set labels
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_zlabel('z')
plt.show()
```

```python
# Create x values
x = np.arange(-10, 10)
# Calculate y values
y = 2 * x + 3

# Create the plot
plt.figure()
plt.plot(x, y, linestyle='--', color='red', marker='o', label='y = x^2')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Plot of y = 2x + 3')
plt.legend()
plt.show()
```
