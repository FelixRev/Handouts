## If-Else Statements

If-Else statements are used to execute a block of code based on a condition.

**Syntax:**
```python
if condition:
    # code to execute if condition is true
else:
    # code to execute if condition is false

````python
age = 18
if age >= 18:
    result = "You are an adult."
else:
    result = "You are a minor."

result
````

## For Loops

For loops are used to iterate over a sequence (like a list, tuple, or string).

**Syntax:**
```python
for item in sequence:
    # code to execute for each item

````python
fruits = ["apple", "banana", "cherry"]
fruit_messages = [f"I like {fruit}." for fruit in fruits]

fruit_messages
````

## While Loops

While loops are used to execute a block of code as long as a condition is true.

**Syntax:**
```python
while condition:
    # code to execute while condition is true

````python
count = 0
count_messages = []
while count < 5:
    count_messages.append(f"Count is: {count}")
    count += 1  # Increment count by 1

count_messages
````
