## Exercise 1

```python
from sklearn.datasets import load_iris
import pandas as pd

from sklearn.datasets import load_iris
import pandas as pd

# Load the Iris dataset
iris_dataset = load_iris()
iris_data = pd.DataFrame(data=iris_dataset.data, columns=iris_dataset.feature_names)

# Create a DataFrame for species
iris_species = pd.DataFrame(data=iris_dataset.target, columns=['species'])

iris_dataframe = pd.DataFrame.join(iris_data, iris_species)

iris_dataframe.head()
```

## Exercise 2

```python
# Check for any missing values
print(iris_dataframe.isnull().sum())

# Get a summary of the dataset
print(iris_dataframe.describe())
```

## Exercise 3

```python
# Fill missing values with the mean of the respective column
iris_dataframe.fillna(iris_data.mean(), inplace=True)

# Drop any rows where the species column is missing
iris_dataframe.dropna(subset=['species'], inplace=True)
```

## Exercise 4

```python
# Calculate average sepal length and petal width
average_sepal_length = iris_dataframe['sepal length (cm)'].mean()
average_petal_width = iris_dataframe['petal width (cm)'].mean()

# Find the species with the maximum petal length
max_petal_length_species = iris_dataframe.loc[iris_data['petal length (cm)'].idxmax()]

# Create a new column for sepal area
iris_dataframe['sepal_area'] = iris_dataframe['sepal length (cm)'] * iris_dataframe['sepal width (cm)']

# Display results
print("Average Sepal Length:", average_sepal_length)
print("Average Petal Width:", average_petal_width)
print("Species with Maximum Petal Length:\n", max_petal_length_species)
print(iris_dataframe.head())
```

## Exercise 5

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
scatter = plt.scatter(iris_dataframe['sepal length (cm)'], iris_dataframe['sepal width (cm)'], c=iris_dataframe['species'], cmap='viridis')
plt.title('Sepal Length vs Sepal Width')
plt.xlabel('Sepal Length (cm)')
plt.ylabel('Sepal Width (cm)')
plt.colorbar(scatter, label='Species')
plt.show()
```
